# Unione dei dati ISTAT e SITUAS

* Decido di prendere l'anno 2024 per fare la merge con l'obbiettivo di trovare il miglior candidato per l'investimento. 

```python
import pandas as pd 
df1 = pd.read_csv("C:/Users/furia/Desktop/Proggetto_Finale/Dati/puliti/incidenti_istat_clean.csv", dtype={'REF_AREA': str})
df2 = pd.read_csv("C:/Users/furia/Desktop/Proggetto_Finale/Dati/puliti/Comuni_Puliti.csv", sep=';', dtype={'Codice Comune (alfanumerico)': str})
df_risultato = pd.merge(df1,df2, left_on='REF_AREA', right_on='Codice Comune (alfanumerico)', how='left')
df_risultato.head(10) 
```

* Info per vedere come sono distribuite le colonne

```python
df_risultato.info()
```

Controllo sui valori nulli, vedo che ci sono degli incidenti che non hanno un comuni e quindi non hanno avuto match

```python
print(df_risultato['Comune'].isna().sum())
```

```python
df_risultato[df_risultato['Comune'].isna()]['REF_AREA'].value_counts().head(10)
```

* Riempio i valori mancanti con 0 capendo che alcuni comuni sono stati dismessi o unificati con altri 

```python
df_risultato = df_risultato.fillna(0)
```
