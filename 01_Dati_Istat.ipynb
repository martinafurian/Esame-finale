# Scaricare dati ISTAT #

* Importe le librerie e tramite request i dati ISTAT sugli incidenti stradali e salvo nellla cartella dati grezzi 

```python
import pandas as pd
import requests
import os
```

```python
os.makedirs("Dati/grezzi", exist_ok=True) 

url = "https://esploradati.istat.it/SDMXWS/rest/data/41_983"
headers = {"Accept": "application/vnd.sdmx.data+csv;version=1.0.0"}

r = requests.get(url, headers=headers)
print(r.status_code)

if r.status_code == 200:
    with open("Dati/grezzi/incidenti_istat.csv", "wb") as f:
        f.write(r.content)
```
